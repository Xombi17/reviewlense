# ReviewLens AI - Google Colab Training Notebook

This notebook runs training in Google Colab using files stored on your Google Drive. It compiles the dataset, trains the models, and provides a convenient download link for the trained model files.


```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Define project root inside Drive – update if your folder name differs
PROJECT_DIR = '/content/drive/MyDrive/mini-project1'

import os, sys
# Add backend package to PYTHONPATH
sys.path.append(os.path.join(PROJECT_DIR, 'backend'))
# Change working directory to backend for relative imports (optional)
os.chdir(os.path.join(PROJECT_DIR, 'backend'))
print('Current working directory:', os.getcwd())

# Verify dataset files exist on Drive
AMAZON_TRAIN_PATH = os.path.join(PROJECT_DIR, 'backend/data/raw/amazon_train.csv')
AMAZON_TEST_PATH = os.path.join(PROJECT_DIR, 'backend/data/raw/amazon_test.csv')
print('Train CSV exists:', os.path.exists(AMAZON_TRAIN_PATH))
print('Test CSV exists:', os.path.exists(AMAZON_TEST_PATH))
```

## 📦 Step 4: Download Trained Model Files
After training completes, the model artefacts are saved under `backend/models/`. This cell zips the directory and triggers a download so you can bring the models back to your local machine.


```python
import shutil, pathlib
models_dir = pathlib.Path(PROJECT_DIR) / 'backend/models'
zip_path = pathlib.Path(PROJECT_DIR) / 'trained_models.zip'
shutil.make_archive(str(zip_path).replace('.zip',''), 'zip', root_dir=models_dir)
print('Model archive created at', zip_path)
# Use Colab utilities to download the zip file
from google.colab import files
files.download(str(zip_path))
```
